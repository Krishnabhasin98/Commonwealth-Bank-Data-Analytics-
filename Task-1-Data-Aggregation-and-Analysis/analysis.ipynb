# Task 1 — Data Aggregation and Analysis
Load the supplied `supermarket_transactions.xlsx` workbook and answer the three business questions using pandas.

```python
import pandas as pd
df = pd.read_excel('supermarket_transactions.xlsx')
df.head()
```

```python
# Dataset overview
print(df.shape)
print(df.dtypes)
print(df['store'].nunique())
```

```python
# 1. Number of apples purchased using cash
apple_cash_qty = df.loc[(df['product_name'].str.lower() == 'apple') & (df['payment_method'].str.lower() == 'cash'), 'quantity'].sum()
print('Cash-purchased apples:', apple_cash_qty)
```

```python
# 2. Total spending on apples
apple_spend = df.loc[df['product_name'].str.lower() == 'apple', 'total_amount'].sum()
print('Total apple spending:', round(apple_spend, 2))
```

```python
# 3. Non-member spending at Bakershire
bakershire_nonmember_spend = df.loc[(df['store'].str.lower() == 'bakershire') & (df['customer_type'].str.lower() == 'non-member'), 'total_amount'].sum()
print('Bakershire non-member spending:', round(bakershire_nonmember_spend, 2))
```
