# Task 3 — Social Media Data Analysis Approach
The workbook contains simulated social-media posts. These findings are illustrative and are not actual Commonwealth Bank social-media activity.

```python
import pandas as pd
df = pd.read_excel('commonwealth_bank_social_media_dataset.xlsx', sheet_name='Social_Media_Posts')
df.head()
```

```python
# Inspect structure and sentiment distribution
df.info()
df['sentiment'].value_counts(dropna=False)
```

```python
# Topics, engagement and response priorities
topic_summary = df.groupby('topic').agg(posts=('topic','size'), total_engagement=('total_engagements','sum'), avg_engagement_rate=('engagement_rate','mean')).sort_values('total_engagement', ascending=False)
display(topic_summary)
print('Posts requiring response:', (df['needs_response'] == True).sum())
```
